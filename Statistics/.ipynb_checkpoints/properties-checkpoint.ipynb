```python
events = []
points = []
```

```python
for cr in tqdm(crs['O']):
    
    Range_start = cr['range_start']
    Range_end = cr['range_end']
    Mean_range = np.mean([Range_start,Range_end])
    Timestamp = cr['timestamp']
    Time = Timestamp % 86400
    Mean_power = np.mean(cr['trail_power_dB_joint'])
    Min_power = np.min(cr['trail_power_dB_joint'])
    Max_power = np.max(cr['trail_power_dB_joint'])
    Pointing = 0
    Trail_length = Range_end-Range_start
    
    event = {"Range_start": Range_start, "Range_end": Range_end, "Mean_range": Mean_range, "Timestamp": Timestamp, 
             "Time": Time, "Mean_power": Mean_power, "Min_power": Min_power, "Max_power": Max_power, "Pointing": Pointing,
             "Trail_length": Trail_length }
    events.append(event)

    for k in np.arange(0,len(cr['trail_power_dB_joint']),1):

        Range = cr['ranges'][cr['range_start_ID'] + k]
        Timestamp = Timestamp
        Time = Time
        Power = cr['trail_power_dB_joint'][k]
        Trail_length = Trail_length

        point = {"Range": Range, "Timestamp": Timestamp, "Time": Time, "Power": Power, "Trail_length": Trail_length }
        points.append(point)
        

for cr in tqdm(crs['P']):
    
    Range_start = cr['range_start']
    Range_end = cr['range_end']
    Mean_range = np.mean([Range_start,Range_end])
    Timestamp = cr['timestamp']
    Time = Timestamp % 86400
    Mean_power = np.mean(cr['trail_power_dB_joint'])
    Min_power = np.min(cr['trail_power_dB_joint'])
    Max_power = np.max(cr['trail_power_dB_joint'])
    Pointing = 1
    Trail_length = Range_end-Range_start
    
    event = {"Range_start": Range_start, "Range_end": Range_end, "Mean_range": Mean_range, "Timestamp": Timestamp, 
             "Time": Time, "Mean_power": Mean_power, "Min_power": Min_power, "Max_power": Max_power, "Pointing": Pointing,
             "Trail_length": Trail_length }
    events.append(event)

    for k in np.arange(0,len(cr['trail_power_dB_joint']),1):

        Range = cr['ranges'][cr['range_start_ID'] + k]
        Timestamp = Timestamp
        Time = Time
        Power = cr['trail_power_dB_joint'][k]
        Trail_length = Trail_length

        point = {"Range": Range, "Timestamp": Timestamp, "Time": Time, "Power": Power, "Trail_length": Trail_length }
        points.append(point)
        
```

```python
df_event = pd.DataFrame(events)
df_points = pd.DataFrame(points)
```

```python
df_event.to_csv("events_2025.csv", index=False)
df_points.to_csv("points_2025.csv", index=False)
```

```python

######### DAY1 ############
pointing='P'
file='trails_jun2023_perpendicular_CR.pickle'
start = datetime(2000, 8, 25, 11, 36, 38)  # YYYY,MM,DD,hh,mm,ss
end = datetime(2035, 8, 27, 0, 0, 0)
crs_temp=[]
with open(os.path.join(directory,file), 'rb') as f:
    print(f'processing {file}')
    not_end=True
    while not_end:
        try:
            obj = pickle.load(f)
            if start<=datetime.fromtimestamp(obj['timestamp'])<=end is not None:
                crs_temp.append(obj)
        except EOFError as e:
            not_end=False
tmin=datetime.fromtimestamp(crs_temp[0]['timestamp'])
tmax=datetime.fromtimestamp(crs_temp[-1]['timestamp'])
print(f'Total: {len(crs_temp)} Cosmic Rays | {tmin} ---> {tmax}')
crs[pointing]+=crs_temp
```

```python
events = []
points = []
```

```python
for cr in tqdm(crs['O']):
    
    Range_start = cr['Range_start']
    Range_end = cr['Range_end']
    Mean_range = np.mean([Range_start,Range_end])
    Timestamp = cr['timestamp']
    Time = Timestamp % 86400
    Mean_power = np.mean(cr['trail_power_dB_joint'])
    Min_power = np.min(cr['trail_power_dB_joint'])
    Max_power = np.max(cr['trail_power_dB_joint'])
    Pointing = 0
    Trail_length = Range_end-Range_start
    
    event = {"Range_start": Range_start, "Range_end": Range_end, "Mean_range": Mean_range, "Timestamp": Timestamp, 
             "Time": Time, "Mean_power": Mean_power, "Min_power": Min_power, "Max_power": Max_power, "Pointing": Pointing,
             "Trail_length": Trail_length }
    events.append(event)
    
    r_list=list(np.linspace(cr['Range_start'],cr['Range_end'],cr['Range_end_ID']-cr['Range_start_ID']+1))
    for k in np.arange(0,len(cr['trail_power_dB_joint']),1):

        Range = r_list[k]
        Timestamp = Timestamp
        Time = Time
        Power = cr['trail_power_dB_joint'][k]
        Trail_length = Trail_length

        point = {"Range": Range, "Timestamp": Timestamp, "Time": Time, "Power": Power, "Trail_length": Trail_length }
        points.append(point)
        

for cr in tqdm(crs['P']):
    
    Range_start = cr['Range_start']
    Range_end = cr['Range_end']
    Mean_range = np.mean([Range_start,Range_end])
    Timestamp = cr['timestamp']
    Time = Timestamp % 86400
    Mean_power = np.mean(cr['trail_power_dB_joint'])
    Min_power = np.min(cr['trail_power_dB_joint'])
    Max_power = np.max(cr['trail_power_dB_joint'])
    Pointing = 1
    Trail_length = Range_end-Range_start
    
    event = {"Range_start": Range_start, "Range_end": Range_end, "Mean_range": Mean_range, "Timestamp": Timestamp, 
             "Time": Time, "Mean_power": Mean_power, "Min_power": Min_power, "Max_power": Max_power, "Pointing": Pointing,
             "Trail_length": Trail_length }
    events.append(event)
    
    r_list=list(np.linspace(cr['Range_start'],cr['Range_end'],cr['Range_end_ID']-cr['Range_start_ID']+1))
    for k in np.arange(0,len(cr['trail_power_dB_joint']),1):

        Range = r_list[k]
        Timestamp = Timestamp
        Time = Time
        Power = cr['trail_power_dB_joint'][k]
        Trail_length = Trail_length

        point = {"Range": Range, "Timestamp": Timestamp, "Time": Time, "Power": Power, "Trail_length": Trail_length }
        points.append(point)


for cr in tqdm(crs['NA']):
    
    Range_start = cr['Range_start']
    Range_end = cr['Range_end']
    Mean_range = np.mean([Range_start,Range_end])
    Timestamp = cr['timestamp']
    Time = Timestamp % 86400
    Mean_power = np.mean(cr['trail_power_dB_joint'])
    Min_power = np.min(cr['trail_power_dB_joint'])
    Max_power = np.max(cr['trail_power_dB_joint'])
    Pointing = None
    Trail_length = Range_end-Range_start
    
    event = {"Range_start": Range_start, "Range_end": Range_end, "Mean_range": Mean_range, "Timestamp": Timestamp, 
             "Time": Time, "Mean_power": Mean_power, "Min_power": Min_power, "Max_power": Max_power, "Pointing": Pointing,
             "Trail_length": Trail_length }
    events.append(event)
    
    r_list=list(np.linspace(cr['Range_start'],cr['Range_end'],cr['Range_end_ID']-cr['Range_start_ID']+1))
    for k in np.arange(0,len(cr['trail_power_dB_joint']),1):

        Range = r_list[k]
        Timestamp = Timestamp
        Time = Time
        Power = cr['trail_power_dB_joint'][k]
        Trail_length = Trail_length

        point = {"Range": Range, "Timestamp": Timestamp, "Time": Time, "Power": Power, "Trail_length": Trail_length }
        points.append(point)
```

```python
df_event = pd.DataFrame(events)
df_points = pd.DataFrame(points)
```
